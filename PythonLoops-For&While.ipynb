#### For in Loop

```python
for a in range(10):
    print('Zaeem Rehman')
```

```python
friends = ['Husain','Umer','Ali','Zaeem','Hashir','Arquam','Faheem']
for friend in friends:
    print(f"Hello {friend}")
```

```python
for n in range(10):
    print(n)
```

```python
# direct iterating over list
for n in friends:
    print(f"Hello {n}")
```

```python
# iterating over list using index
for n in range(6):
    print(n, friends[n])
```

```python
for n in range(len(friends)):
    print(n, friends[n])
```

```python
# to print the list items in reverse order we will reverse the list first
for n in friends[::-1]:
    print(n)
```

```python
for n,s in enumerate(friends[::-1]):
    print(n,s)
```

```python
# Enumerate(): This allows you to track both the index and the value of items during a loop without needing a manual counter variable.
```

#### Continue , break

```python
students = ['Husain','Umer','Ali','Zaeem','Hashir','Arquam','Faheem']
for s in students:
    print(f"Respected Mr {s}, Today is your Off")
```

```python
for s in students:
    if len(s)<6:
            print(f"Respected Mr {s}, Today is your Off")
    else:
            print(f"Respected Mr {s}, You have to come Today")
```

```python
for s in students:
    if s=="Arquam":
        break
    else:
        print(s)
      
```

# While Loop

```python
friends = []
while True:
        name = input("Enter guest name press quit to exit").lower()
        if name=="quit":
                break
        friends.append(name)
friends
```
