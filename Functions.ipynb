```python
def text_analyzer(txt):
    length = len(txt)
    words = len(txt.split(" "))
    sentence = len(txt.split("."))
    paragraphs = len(txt.split("\n"))
    vow = "aeiouAEIOU"
    punctuation = '!"#$%&\'()*+,-./:;<=>?@[\\]^_`{|}~'
    alphabets = 0
    spaces = 0
    numbers = 0
    lower_case = 0
    Upper_case = 0
    Vowels = 0
    punctuations = 0
    others = 0
    show_others = ' '
    for char in txt:
        if char.isalpha():
            alphabets += 1
            if char.isupper():
                Upper_case += 1
            else:
                lower_case += 1
            if char in vow:
                Vowels += 1
        elif char.isspace():
            spaces += 1
        elif char.isnumeric():
            numbers += 1
        elif char in punctuation:
            punctuations += 1
        else:
            others += 1
            show_others += char

    print(f"""
      Text Length : {length}
      Words Length : {words}
      Length of Sentences : {sentence}
      Length of Paragraphs : {paragraphs}
      Upper Case : {Upper_case}
      Lower Case : {lower_case}
      Spaces are : {spaces}
      Numbers are : {numbers}
      Alphabets are : {alphabets}
      Vowels : {Vowels}
      Punctuations : {punctuations}
      Others : {others}
      others are : {show_others}

    """)

    
```

```python
def shopping():
    cart = {}
    while True:
        requiredItem = input("Enter the item you want to buy or press q to exit").lower()
        if requiredItem == "q":
            break
        else:
            for a in groceryShop:
                if a['item'] == requiredItem:
                    print(f"The price for {requiredItem} is {a['price']}")
                    quantity = float(input("Enter required quantity: "))
                    if a['inventory'] >= quantity:
                        cart[requiredItem] = a['price'] * quantity
                        a['inventory'] -= quantity    
                    else:
                        print(f"{requiredItem} is not available in your required quantity")   
```
