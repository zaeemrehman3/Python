```python
type(nums)
```

```python
bnums = {}  # empty is not set
type(bnums)
```

```python
nums
```

```python
setA = {1, 2, 3, 4, 5, 6, 7, 100}
setB = {1,6,8,8,58,8,7,2,8,100}
setA.difference(setB)
```
