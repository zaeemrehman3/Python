```python
bill_Amount = int(input("Enter your bill Amount"))
if bill_Amount>50000:
    print("10% discount")
else:
    print("Sorry,no discount")
```

```python
billAmount = int(input("Enter your bill Amount:"))
if billAmount > 50000:
    discount = billAmount*0.30
    print(f"You have got discount of amount: {discount} on bill of amount: {billAmount}")
else:
    print(f"Sorry you didn't got any discount on bill of amount: {billAmount}")
```

```python
Salary = int(input("Enter your Salary"))
if Salary >= 600000:
    abovesalaryTax = Salary*0.05
    abovenetamount = Salary-abovesalaryTax
    print(f"The 5% Tax Deduction Amount is Rs: {abovesalaryTax} from your salary of Rs: {Salary} now Your Current Salary is Rs: {abovenetamount}")
else:
    lowsalaryTax = Salary*0.02
    lownetamount = Salary-lowsalaryTax
    print(f"The 2% Tax Deduction Amount is Rs: {lowsalaryTax} from your salary of Rs: {Salary} now Your Current Salary is Rs: {lownetamount}")

    
    
```

```python
Salary = int(input("Enter your Salary"))
if Salary >= 2400000:
    taxAmount = Salary*.15
elif Salary >= 1200000:
    taxAmount = Salary*.10
elif Salary >= 600000:
    taxAmount = Salary*.05
else:
    taxAmount = Salary*0

print(f"Your Salary is Rs: {Salary}")
print(f"Your Tax Amount is Rs: {taxAmount}")
print(f"Your Net Salary per Annum is Rs: {Salary-taxAmount}")

```
